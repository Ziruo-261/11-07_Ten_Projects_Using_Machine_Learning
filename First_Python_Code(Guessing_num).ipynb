```python
import random
secret = random.randint(1, 10)
guess = 0
attempts_left = 5
while guess != secret:
  if attempts_left > 0:
    print("number of guesses left:", attempts_left)
    guess = int(input("Guess a number (1-10): "))
    attempts_left -= 1
    if guess < secret:
      print("Too low!")
    elif guess > secret:
      print("Too high!")
  else:
    print("Sorry, you have run out of attempts. :( Try again next time!")
    break
if attempts_left >= 0 and guess == secret:
  print("You got it!")
```
